# **Exploratory Data Analysis**



## **Gender**

```python
df.columns
```

```python
# plotting a bar chart for Gender and it's count
ax=sns.countplot(x='Gender',data=df)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
# Sales Distribution by Gender
sales_gen = df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
colors = sns.color_palette('pastel')[0:len(sales_gen)]
plt.figure(figsize=(6, 6))
plt.pie(sales_gen['Amount'], labels=sales_gen['Gender'], colors=colors, autopct='%1.1f%%', startangle=140)
plt.axis('equal')
plt.show()
```

```python
# plotting a bar chart for gender vs total amount
df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
```

```python
sales_gen=df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.set(rc={'figure.figsize':(7,5)})
sns.barplot(x='Gender', y='Amount', data=sales_gen)
```

*The graphs show that the majority of buyers are female, and their purchasing power is higher than that of males.*

### **Marital Status**

```python
ax = sns.countplot(data = df, x = 'Marital_Status')

sns.set(style="whitegrid", rc={'figure.figsize':(20,5)})
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_marital = df.groupby(['Marital_Status', 'Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.set(style="whitegrid", rc={'figure.figsize':(6,5)})
sns.barplot(data = sales_marital, x = 'Marital_Status',y= 'Amount', hue='Gender')
```

*The graphs show that most buyers are married women, and they have higher purchasing power.*

### **Occupation**

```python
ax = sns.countplot(data = df, x = 'Occupation', hue='Occupation', palette=palette, legend=False)
sns.set(rc={'figure.figsize':(25,5)})
palette = sns.color_palette("husl", len(df['Occupation'].unique()))

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_occupation = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.set(rc={'figure.figsize': (20, 5)})
palette = sns.color_palette("husl", len(sales_occupation))
ax = sns.barplot(data=sales_occupation, x='Occupation', y='Amount', hue='Occupation')
```

From above graphs we can see that most of the buyers are working in IT, Healthcare and Aviation sector

### ***Product Category***

```python
sns.set(rc={'figure.figsize':(25,5)})
palette = sns.color_palette("husl", len(df['Product_Category'].unique()))
ax = sns.countplot(data = df, x = 'Product_Category', hue='Product_Category', palette=palette, legend=False)

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_product = df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)
sns.set(rc={'figure.figsize': (20, 5)})
palette = sns.color_palette("husl", len(sales_product))
ax = sns.barplot(data=sales_product, x='Product_Category', y='Amount',hue='Product_Category', palette=palette, legend=False)
```

```python
sales_product = df.groupby(['Product_ID'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)
sns.set(style="whitegrid", rc={'figure.figsize':(20,5)})
palette = sns.color_palette("husl", len(sales_product))
sns.barplot(data = sales_product, x = 'Product_ID',y= 'Orders',hue='Product_ID', palette=palette, legend=False)
```
